```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Introduction to the COUNT() aggegate fuction.
```

```python
#1 Count the total number of employees in the employees table.
run_sql("""SELECT COUNT(*) AS total_employees
FROM employees;""")
```

```python
#2 Count the total number of departments.
run_sql("""SELECT COUNT(*) AS total_departments
FROM departments;""")
```

```python
#3 Count the total number of products.
run_sql("""SELECT COUNT(*) AS total_products
FROM products;""")
```

```python
#4 Count the total number of customers.
run_sql("""SELECT COUNT(*) total_customers
FROM customers;""")
```

```python
#5 Count the total number of orders.
run_sql("""SELECT COUNT(*) AS total_orders
FROM ORDERS;""")
```

```python
# Bouns Challege
run_sql("""SELECT COUNT(email) AS total_emails
FROM employees;""")
```
