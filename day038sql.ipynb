```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Using Multiple Aggregate Funtions Together
```

```python
#1 Display the total number of employees and the minimum and miximum salary.
run_sql("""SELECT COUNT(*) AS total_employees,
                MIN(salary) AS minimum_salary,
                MAX(salary) AS maximum_salary
         FROM employees;""")
```

```python
#2 Show the total number of products along with the minimum and maiximum product price.
run_sql("""SELECT COUNT(*) AS total_products,
         MIN(price) AS minimum_price,
         MAX(price) AS maximum_price
  FROM  products;""")
```

```python
#3 Display the total number of orderes, total orders, total ordered quantity, and average order quantity.
run_sql("""SELECT COUNT(*) AS total_orders,
      SUM(quantity) AS total_ordered_quantity,
      AVG(quantity) AS averge_order_quantity
 FROM orders;""")
```

```python
#4 For products in the Electronics category, display the total number of products, minimum price, maximum price, and average price.
run_sql("""SELECT COUNT(*) AS total_product,
    MIN(price) AS minimum_price,
    MAX(price) AS maximum_price,
    AVG(price) AS average_price
FROM products
WHERE category = 'Electronis'""")
```

```python
#5 For employees in department_id = 2, display the total number of employees, total salary, average salary, minimum salary, and maximum salary.
run_sql("""SELECT COUNT(*) AS total_employees,
         SUM(salary) AS total_salary,
         AVG(salary) AS average_salary,
         MIN(salary) AS minimum_salary,
         MAX(salary) AS maximum_salary
   FROM employees
   WHERE department_id = 2;""")
```

```python
# Bonus Challege
run_sql("""SELECT COUNT(*) AS total_orders,
      MIN(order_date) AS earliest_order_date,
      MAX(order_date) AS latest_order_date
 FROM orders;""")
```
