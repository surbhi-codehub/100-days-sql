```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
#1 Display employees whose employee_id is BETWEEN 1 AND 5.
run_sql("""SELECT *
FROM employees
WHERE employee_id BETWEEN 1 AND 5;""")
```

```python
#2 Display customers whose city is 'Bangalore' or 'Mumbai' using the IN operator.
run_sql("""SELECT *
FROM customers
WHERE city IN ('Bangalore', 'Mumbai');""")
```

```python
#3 Display products whose category is NOT 'Electronic'.
run_sql("""SELECT *
FROM products
WHERE category NOT IN ('Electronics');""")
```

```python
#4 Display orders where customer_id is IN(1,2,3) and order_id is BETWEEN 1 AND 20.
run_sql("""SELECT *
FROM orders
WHERE customer_id IN (1,2,3)
AND order_id BETWEEN 1 AND 20;""")
```

```python
#5 Display employee_id,first_namel, and last_name for employees whose department_id is IN(1,2) and employee_id is BETWEEN 2 AND 8.
run_sql("""SELECT employee_id, first_name, last_name
FROM employees
WHERE department_id IN (1,2)
AND employee_id BETWEEN 2 AND 8;""")
```

```python
# Bonus Challenge
# Write one SQL query that combines at leat two filtering operators.
run_sql("""SELECT *
FROM employees
WHERE department_id IN (1,2)
AND employee_id BETWEEN 1 AND 10
AND first_name IS NOT NULL;""")
```
