```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: GROUP BY &HAVING - Final practice.
```

```python
#1
run_sql("""SELECT department_id, COUNT(*) AS total_employees
FROM employees
GROUP BY department_id
HAVING COUNT(*) >= 2;""")
```

```python
#2
run_sql("""SELECT category, AVG(price) AS average_price
FROM products
WHERE stock_quantity > 20
GROUP BY category
HAVING AVG(price) > 1000;""")
```

```python
#3
run_sql("""SELECT customer_id,
         COUNT(*) AS total_orders,
         SUM(quantity) AS total_order_quantity
         FROM orders
         GROUP BY customer_id
         HAVING SUM(quantity) > 15;""")
```

```python
#4
run_sql("""SELECT department_id,
         MIN(salary) AS minimum_salary,
         MAX(salary) AS maximum_salary,
         AVG(salary) AS average_salary

        FROM employees
        WHERE hire_date > '2022-01-01'
        GROUP BY department_id
        HAVING AVG(salary) > 60000;""")
```

```python
#5
run_sql("""SELECT category,
                  SUM(stock_quantity) AS total_stock_quantity,
                  AVG(price) AS average_price,
                  MIN(price) AS minimum_price,
                  MAX(price) AS maximum_price

          FROM products
          WHERE price > 500
          GROUP BY category
          HAVING SUM(stock_quantity) > 200;""")
```

```python
# Bonus Challege
run_sql("""SELECT customer_id,
       MAX(quantity) AS maximum_order_quantiy

       FROM orders
       GROUP BY customer_id
       HAVING COUNT(*) > 2;""")
```
