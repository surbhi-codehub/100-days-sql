```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Using the single-character wildcard_with LIKE.
```

```python
#1 Display customers whose customer_name starts with 'A' and has exactly one more charcter.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'A_';""")
```

```python
#2 Display employees whose first_name has exactly five characters.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE'___';""")
```

```python
#3 Display product_name starts with 'S' followed by exactly one charcter and then any remaining charcters.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE 'S_%';""")
```

```python
#4 Display departments whose department_name has 'e' as the second charcter.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE '_e%';""")
```

```python
#5 Find customers whose city start with 'B', has exactly one unknown charcter, and then any remaining characters.
run_sql("""SELECT *
FROM customers
WHERE city LIKE 'B_%'""")
```

```python
# Bonus Challenge:
run_sql("""SELECT *
FROM employees
WHERE last_name LIKE 'J_%';""")
```
