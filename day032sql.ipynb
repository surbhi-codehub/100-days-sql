```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: COUNT() with Conditions
```

```python
#1 Count how many employees belong to the Sales department.
run_sql("""SELECT COUNT(*)
FROM employees
WHERE department_id = 1;""")
```

```python
#2 Count how many products have a price greater than 100.
run_sql("""SELECT COUNT(*)
FROM products
WHERE price > 100;""")
```

```python
#3 Count how many customers live in Delhi.
run_sql("""SELECT COUNT(*)
FROM customers
WHERE city = 'Delhi';""")
```

```python
#4 Count the number of orders where the quantity is greater than 3.
run_sql("""SELECT COUNT(*)
FROM orders
WHERE quantity > 3;""")
```

```python
#5 Count how many employees earn a salary greater than 60000 and belong to department 2.
run_sql("""SELECT COUNT(*)
FROM employees
WHERE salary > 60000
AND department_id = 2.""")
```

```python
# Bonus Challenge
run_sql("""SELECT COUNT(*)
FROM products
WHERE stock_quantity < 20;""")
```
