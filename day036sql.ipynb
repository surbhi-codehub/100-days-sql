```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
#1 Find the lowest product price in the in the poducts table.
run_sql("""SELECT MIN(price)
FROM products;""")
```

```python
#2 Find the minimum salary amonge all employees.
run_sql("""SELECT MIN(salary)
FROM employees;""")
```

```python
#3 Find the earliest order_date in the orders table.
run_sql("""SELECT MIN(order_date)
FROM orders;""")
```

```python
#4 Find the mininmum stock quantity for products in the Electronics category.
run_sql("""SELECT MIN(stock_quantity)
FROM products
WHERE category = 'Electronics';""")
```

```python
#5 Find the earliest hire date of employees whose  department_id is 3.
run_sql("""SELECT MIN(hire_date)
FROM employees
WHERE department_id = 3;""")
```

```python
# Bonus Challege
run_sql("""SELECT MIN(quantity)
FROM orders
WHERE quantity > 1;""")
```
