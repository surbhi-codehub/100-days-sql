```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Using LIKE to find text that contain specific characters.
```

```python
#1 Display all customers whose customer_name contains the letters 'an'.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE '%an%';""")
```

```python
#2 Display all employees whose first_name contains the letter 'a'.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE '%a%';""")
```

```python
#3 Display all products whose product_name contains the word 'pro'.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE '%pro%';""")
```

```python
#4 Display all department whose department_name contains the letter 'e'.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE '%e%';""")
```

```python
#5 Display all customers whose city contains the letter 'o'.
run_sql("""SELECT *
FROM customers
WHERE city LIKE '%O%';""")
```
