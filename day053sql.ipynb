```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Left JOIN - Practical Reports
```

```python
#1.
run_sql("""SELECT
c.customer_id,
c.first_name,
c.last_name,
COUNT(o.order_id) AS total_orders
FROM customers c
LEFT JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.first_name, c.last_name;""")
```

```python
#2.
run_sql("""SELECT
          d.department_id,
          d.department_name,
          AVG(e.salary) AS average_salary
          FROM departments d
          LEFT JOIN employees e
           ON d.department_id = e.department_id
          GROUP BY d.department_id, d.department_name;""")
```

```python
#4.
run_sql("""SELECT
    p.product_id,
    p.product_name,
    COALESCE(SUM(o.quantity), 0) AS total_quantity_ordered
FROM products p
LEFT JOIN orders o
    ON p.product_id = o.product_id
GROUP BY p.product_id, p.product_name;""")
```

```python
#5.
run_sql("""SELECT
 c.customer_id,
 c.first_name,
 COUNT(DISTINCT o.order_id ) AS total_orders,
 COALESCE(SUM(o.quantity), 0) AS total_quantity_ordered
 FROM customers c
 LEFT JOIN orders o
 ON o.customer_id = o.customer_id
 GROUP BY c.customer_id, c.first_name, c.last_name;""")
```

```python
# Bonus Challenge
run_sql("""SELECT
d.department_id,
d.department_name,
MAX(e.salary) AS highest_salary
FROM departments d
LEFT JOIN employees e
ON d.department_id = e.department_id
GROUP BY d.department_id, d.department_name;""")
```
