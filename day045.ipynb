```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: GROUP BY & HAVING - Practice Day
```

```python
#1.
run_sql("""SELECT department_id, COUNT(*) AS  total_employees
FROM  employees
GROUP BY department_id
HAVING COUNT(*) >= 2;""")
```

```python
#2.
run_sql("""SELECT category, COUNT(*) AS total_products
FROM products
WHERE price > 500
GROUP BY category
HAVING COUNT(*) > 2;""")
```

```python
#3.
run_sql("""SELECT customer_id, SUM(quantity) AS total_quantity
FROM orders
WHERE quantity >= 2
GROUP BY customer_id
HAVING SUM(quantity) >= 8;""")
```

```python
#4.
run_sql("""SELECT department_id, AVG(salary) AS average_salary
FROM employees
WHERE salary > 40000
GROUP BY department_id
HAVING AVG(salary) > 60000;""")
```

```python
#5.
run_sql("""SELECT category, SUM(stock_quantity) AS total_stock
FROM products
WHERE price > 1000
GROUP BY category
HAVING SUM(stock_quantity) > 150;""")
```

```python
# Bonus Challenge
run_sql("""SELECT customer_id,
        COUNT(*) AS total_orders,
        MAX(quantity) AS maximum_order_quantity
        FROM orders
        GROUP BY customer_id
        HAVING COUNT(*) > 1;""")
```
