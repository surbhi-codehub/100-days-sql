```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Aggregate Funtion - Final Practice
```

```python
#1 Display the total numbers of employees and the highest salary.
run_sql("""SELECT COUNT(*) AS total_employees,
           MAX(salary) AS highest_salary
           FROM employees;""")
```

```python
#2 Find the total stock quantity and the average product price from the products table.
run_sql("""SELECT SUM(stock_quantity) AS total_stock_quantity,
             AVG(price) AS average_price
             FROM products;""")
```

```python
#3 Display the total number of orders, the minimum quantity , and the maximum quantiy ordered.
run_sql("""SELECT COUNT(*) AS total_orders,
         MIN(quantity) AS minimum_quantity,
         MAX(quantity) AS maximum_quantity
         FROM orders;""")
```

```python
#4
run_sql("""SELECT COUNT(*) AS total_products,
                 SUM(stock_quantity) AS total_stock_quantity,
                 AVG(price) AS averge_price,
                 MIN(price) AS minimum_price,
                 MAX(price) AS highest_price
        FROM products
        WHERE category = 'Electronics';""")
```
