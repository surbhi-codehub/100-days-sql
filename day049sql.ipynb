```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# INNER JOIN with Multiple Tables
```

```python
#1 Display each order along with the customer name.
run_sql("""SELECT orders.*, customers.first_name
        FROM orders
        INNER JOIN customers
        ON orders.customer_id = customers.customer_id;""")
```

```python
#2. Display each product together with its supplier name.

# no supplier table.
```

```python
#3 Display each employee along with the department name.
run_sql("""SELECT employees.*, departments.department_name
FROM employees
INNER JOIN departments
ON employees.department_id = departments.department_id;""")
```

```python
#4 Display each order with the customer name and product name.
run_sql("""SELECT orders.*, customers.first_name, products.product_name
        FROM orders
        INNER JOIN customers
        ON orders.customer_id = customers.customer_id
        INNER JOIN products
        ON orders.product_id = products.product_id;""")
```

```python
#5 Display each customer with the total number of orders placed using INNER JOIN and GROUP BY.
run_sql("""SELECT customers.*, COUNT(orders.order_id) AS total_orders
        FROM customers
        INNER JOIN orders
        ON customers.customer_id = orders.customer_id
        GROUP BY customers.first_name;""")
```

```python
# Bonus Challege
run_sql("""SELECT d.department_name, AVG(e.salary) AS average_salary
        FROM departments d
        INNER JOIN employees e
        ON d.department_id = e.department_id
        GROUP BY d.department_name;""")
```
