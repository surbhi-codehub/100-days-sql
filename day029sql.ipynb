```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Mixed practice LIKE,%, and_wildcards.
```

```python
#1 Display customers whose customer_name starts with the letter 'S'.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'S%';""")
```

```python
#2 Display employees whsoe first_name ends with the letter 'n'.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE '%n';""")
```

```python
#3 Display products whose product_name contains the letters 'pro'.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE '%pro%';""")
```

```python
#4 Display departments whose department_name has exactly one character before the letter 'e'.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE '_e%';""")
```

```python
#5 Display customers whose city starts with'B' and ends with 'e'.
run_sql("""SELECT *
FROM customers
WHERE city LIKE 'B%e';""")
```

```python
# Bonus Challege:
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE 'J%n_';""")
```
