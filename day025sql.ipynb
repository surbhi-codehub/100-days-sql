```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Using LIKE to find that end with specific letters.
```

```python
#1 Display all customers whose customer_name ends with the letter 'n'.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE '%n';""")
```

```python
#2 Display all employees whose first_naem ends with the letter 'a'.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE '%a';""")
```

```python
#3 Display all products whose product_name ends with the letter 'r'.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE '%r';""")
```

```python
#4 Display all department whose department_naem ends with the letter 's'.
run_sql("""SELECT *
FROM  departments
WHERE department_name LIKE '%s';""")
```

```python
#5 Dsiplay all cutomers whose city ends with the letter'i'.
run_sql("""SELECT *
FROM customers
WHERE city LIKE '%i';""")
```
