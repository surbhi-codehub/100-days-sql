```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Combining WHERE, GROUP BY, and HAVING.
```

```python
# 1.
run_sql("""SELECT department_id, COUNT(*) AS total_employees
FROM employees
WHERE salary > 30000
GROUP BY department_id
HAVING COUNT(*) > 2;""")
```

```python
# 2
run_sql("""SELECT category, AVG(price) AS average_price
FROM products
WHERE stock_quantity > 10
GROUP BY category
HAVING AVG(price) > 1000;""")
```

```python
# 3.
run_sql("""SELECT customer_id, SUM(quantity) AS total_quantity
FROM orders
WHERE quantity > 1
GROUP BY customer_id
HAVING SUM(quantity) > 5;""")
```

```python
# 4.
run_sql("""SELECT department_id, MAX(salary) AS maximum_salary
FROM employees
WHERE hire_date > '2022-01-01'
GROUP BY department_id
HAVING MAX(salary) > 70000;""")
```

```python
# 5.
run_sql("""SELECT category, SUM(stock_quantity) AS total_stock
FROM products
WHERE price > 500
GROUP BY category
HAVING SUM(stock_quantity) > 100;""")
```

```python
# Bonus Challenge
run_sql("""SELECT customer_id, COUNT(*) AS total_orders
        FROM orders
        WHERE order_date > '2023-01-01'
        GROUP BY customer_id
        HAVING COUNT(*) > 1;""")
```
