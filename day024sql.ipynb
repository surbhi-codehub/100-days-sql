```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Finding text using the LIKE operator.
```

```python
#1 Display all customers whose customer_name starts with  the letter 'A'.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'A%';""")
```

```python
#2 Display all employees whose first_name starts with the letter 'J'.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE 'J%';""")
```

```python
#3 Display all products whos products_name starts with the letter 'S'.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE 'S%';""")
```

```python
#4 Display all department whose department_name starts with the letter 'H'.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE 'H%';""")
```

```python
#5 Display all customers whose city starts with  the letter 'B'.
run_sql("""SELECT *
FROM customers
WHERE city LIKE 'B%';""")
```
