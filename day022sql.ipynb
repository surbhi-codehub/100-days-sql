```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
#1 Display all department sorted by department_name in descending order.
run_sql("""SELECT*
FROM departments
ORDER BY department_name DESC;""")
```

```python
#2 Display employees sorted by first_name indescending order.
run_sql("""SELECT *
FROM employees
ORDER BY first_name DESC;""")
```

```python
# Display products sorted by price from higest to lowest.
run_sql("""SELECT *
FROM products
ORDER BY price DESC;""")
```

```python
#4 Display customers sorted by customer_name inreverse alphabetical order.
run_sql("""SELECT *
FROM customers
ORDER BY first_name DESC;""")
```

```python
#5 Display orders sorted by order_date from newest to oldest.
run_sql("""SELECT *
FROM orders
ORDER BY order_date DESC;""")
```
