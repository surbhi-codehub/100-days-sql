```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Learn how to filter grouped result after using GROUP BY.
```

```python
#1 Display each department_id and the number of employees. Show only department that have more than 2 employees.
run_sql("""SELECT department_id, COUNT(*) AS employee_count
FROM employees
GROUP BY department_id
HAVING COUNT(*) > 2;""")
```

```python
#2 Display each product category and the total number of products. Show only categories with more than 5 products.
run_sql("""SELECT category, COUNT(*) AS product_count
FROM products
GROUP BY category
HAVING COUNT(*) > 5;""")
```

```python
#3 Display each customer and the total number of orders. Show only customer who place more than 1 order.
run_sql("""SELECT customer_id, COUNT(*) AS  order_count
FROM orders
GROUP BY customer_id
HAVING COUNT(*) > 1;""")
```

```python
#4 Display each department _id and  the average salary. Show only department where the average salary is greater than 50000.
run_sql("""SELECT department_id, AVG(salary) AS average_salary
FROM employees
GROUP BY department_id
HAVING AVG(salary) > 5000;""")
```

```python
#5 Dispaly each product category with total stock quantity.Show only categories where the total stock quantity than 100.
run_sql("""SELECT category, SUM(stock_quantity) AS total_stock
FROM products
GROUP BY category
HAVING SUM(stock_quantity) > 100;""")
```

```python
# Bonus Challege
run_sql("""SELECT customer_id, SUM(quantity) AS total_ordered_quantity
FROM orders
GROUP BY customer_id
HAVING SUM(quantity) > 10;""")
```
