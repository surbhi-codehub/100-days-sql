```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Practice with Aggregate Functions.
```

```python
#1 Display the total number of products and the avergae product price.
run_sql("""SELECT COUNT(*) AS TOTAL_PRODUCTS,
        AVG(price) AS average_price
        FROM products;""")
```

```python
#2 Find the total qunatity order and the maximum quantity ordered from the orders table
run_sql("""SELECT SUM(quantity) AS total_quantity,
        MAX(quantity) AS maximum_quantity
        FROM orders;""")
```

```python
#3 Display the minimum , maximum and average salary of all employees.
run_sql("""SELECT MIN(salary) AS minimum_salary,
                  MAX(salary) AS maximum_salary,
                  AVG(salary) AS average_salary
           FROM employees;""");
```

```python
#4 For the products in Electronics category , display the total number of products, total stock quantity , and average product price.
run_sql("""SELECT COUNT(*) AS total_products,
                 SUM (stock_quantity) AS total_stock,
                  AVG(price) AS average_price
            FROM products
            WHERE category = 'Electronic';""")
```

```python
#5
run_sql("""SELECT COUNT(*) AS total_employees,
                  SUM(salary) AS total_salary,
                  AVG(salary) AS average_salary,
                  MIN(salary) AS minimum_salary,
                  MAX(salary) AS maximum_salary
         FROM employees
         WHERE department_id = 1;""")
```

```python
# Bonus Challege
run_sql("""SELECT COUNT(*) AS  total_orders,
                   MIN(quantity) AS minimum_quantity,
                   MAX(quantity) AS maximum_quantity,
                   AVG(quantity) AS average_quantity
        FROM orders;""")
```
