```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Learn how to find the largest value in a numeric or data column using MAX().
```

```python
#1 Find the highest salary among all employees.
run_sql("""SELECT MAX(salary) AS highest_salary
FROM  employees;""")
```

```python
#2 Find the maximum product price.
run_sql("""SELECT MAX(price) AS maximum_price
FROM products;""")
```

```python
#3 Find the latest order date in the orders table.
run_sql("""SELECT MAX(order_date) AS latest_order_date
FROM orders;""")
```

```python
#4 Find the highest stock quantity for products in the Electronics category.
run_sql("""SELECT MAX(stock_quantity) AS highest_stock_quantity
FROM products
WHERE category = 'Electronics';""")
```

```python
#5 Find the highest salary of employees whose department_id is 2.
run_sql("""SELECT MAX(salary) AS higest_salary
FROM employees
WHERE department_id = 2;""")
```

```python
# Bonus Challege
run_sql("""SELECT MAX(hire_date) AS latest_hire_date
FROM employees
WHERE salary > 50000;""")
```
