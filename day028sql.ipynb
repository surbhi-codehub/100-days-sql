```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Combining LIKE pattens using % and_.
```

```python
#1 Display customer whose customer_name starts with 'A'.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'A%';""")
```

```python
#2 Display employees whose first_name contains the contains the letter 'a'.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE '%a%';""")
```

```python
#3 Display product whose product_name starts with 'S' and ends with 'r'.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE 's%r';""")
```

```python
#4 Display department whose department_name has 'e' as the second character.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE '_e%';""")
```

```python
#5 Display customers whose city starts with 'B' and contains the letter 'o' anywhere after the first charcter.
run_sql("""SELECT *
FROM customers
WHERE city LIKE 'B%o%';""")
```

```python
# Bonus Challenge
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'A_%';""")
```
