```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: COUNT () with WHERE
```

```python
#1 Count the total number of customers.
run_sql("""SELECT COUNT(*) AS total_customers
FROM customers;""")
```

```python
#2 Count how many products belong to the Electrnics category.
run_sql("""SELECT COUNT(*) AS electronics_products
FROM products
WHERE category = 'Electronics';""")
```

```python
#3 Count how many employees work in the Sales department using the department_id.
run_sql("""SELECT COUNT(*) AS sales_employees
FROM employees
WHERE department_id = (
   SELECT department_id
   FROM departments
   WHERE department_name = 'Sales'
   );""")
```

```python
#4 Count the number of orders placed where the quantity is greater than 3.
run_sql("""SELECT COUNT(*) AS orders_quantity_greater_than_3
FROM orders
WHERE quantity > 3;""")
```

```python
#5 Count how many employees earn a salary greater than 60000.
run_sql("""SELECT COUNT(*) AS employees_salary_greater_than_60000
FROM employees
WHERE salary > 60000;""")
```

```python
# Bonus Challege
run_sql("""SELECT COUNT(*) AS matching_customers
FROM customers
WHERE city IN (
    SELECT city
    FROM employees
    );""")
```
