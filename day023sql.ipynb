```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Sorting data using multiple columns with ORDER BY.
```

```python
#1
run_sql("""SELECT first_name, last_name
FROM employees
ORDER BY first_name ASC, last_name ASC;""")
```

```python
#2
run_sql("""SELECT *
FROM products
ORDER BY CATEGORY ASC, product_name ASC;""")
```

```python
#3
run_sql("""SELECT *
FROM customers
ORDER BY city ASC, first_name ASC;""")
```

```python
#4
run_sql("""SELECT *
FROM orders
ORDER BY customer_id ASC, order_date ASC;""")
```

```python
#5
run_sql("""SELECT *
FROM employees
ORDER BY department_id ASC, first_name ASC;""")
```

```python
# Bonus Challege
run_sql("""SELECT *
FROM employees
ORDER BY department_id ASC, salary DESC;""")
```
