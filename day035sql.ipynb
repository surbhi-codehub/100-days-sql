```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Learn how to calculate the average value of numeric data using AVG().
```

```python
# Find the average salary of all employees.
run_sql("""SELECT AVG(salary) AS average_salary
FROM employees;""")
```

```python
#2 Find the avergare price of all products.
run_sql("""SELECT AVG(price) AS average_price
FROM products;""")
```

```python
#3 Find the average quantity orderd in the orders table.
run_sql("""SELECT AVG(quantity) AS average_quantity
FROM orders;""")
```

```python
#4 Find the average salary of employees whose deaprtment_id is 2.
run_sql("""SELECT AVG(salary)
FROM employees
WHERE department_id = 2;""")
```

```python
#5 Find the average price of products in the Electronics category.
run_sql("""SELECT AVG(price) AS average_price
FROM products
WHERE category = 'Electronic';""")
```

```python
# Bonus Challege
run_sql("""SELECT AVG(stock_quantity) AS average_stock_quantity
FROM products
WHERE stock_quantity > 50;""")
```
