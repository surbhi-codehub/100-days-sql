```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Business-style filtering practice using WHERE operator.
```

```python
#1 Display all employees whose employee_id is BETWEEN 2 AND 6.
run_sql("""SELECT *
FROM employees
WHERE employee_id BETWEEN 2 AND 6;""")
```

```python
#2 Display customer whosecity inIN('Delhi',Mumbai').
run_sql("""SELECT *
FROM customers
WHERE city IN ('Delhi', 'Mumbai');""")
```

```python
#3 Display products whose category whose categoryis NOT 'Furniture' AND price is BETWEEN 500 AND 5000.
run_sql("""SELECT *
FROM products
WHERE category NOT IN ('Furniture')
 AND price BETWEEN 500 AND 5000;""")
```

```python
#4 Display orders where customer_id IN (1,2,3) AND employee_id IN (1,2).
run_sql("""SELECT *
FROM orders
WHERE customer_id IN (1,2,3)
 AND product_id IN (1,2);""")
```

```python
#5 Display employee_id, first_name, and last_name for employees whose department_id is 2 OR 3, but whose employee_id is NOT BETWEEN 1 AND 2.
run_sql("""SELECT employee_id, first_name, last_name
FROM employees
WHERE (department_id =2 OR department_id =3)
 AND employee_id NOT BETWEEN 1 AND 2;""")
```
