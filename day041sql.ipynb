```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Learn how to group rows with the same value and calculate summaries for each group.
```

```python
#1 Display the number of employees in each department_id.
run_sql("""SELECT department_id, COUNT(*) AS employees_count
FROM employees
GROUP BY department_id;""")
```

```python
#2 Display the number of products in each product category.
run_sql("""SELECT category, COUNT(*) AS product_count
FROM products
GROUP BY category;""")
```

```python
#3 Display the stock quantity for each product category.
run_sql("""SELECT category, (stock_quantity) AS  total_stock
FROM products
GROUP BY category;""")
```

```python
#4 Display the average salary for each department_id.
run_sql("""SELECT department_id, AVG(salary) AS aveage_salary
FROM employees
GROUP BY  department_id;""")
```

```python
#5 Display the total number of orders and the total ordered quantity for each customer.
run_sql("""SELECT customer_id,
        COUNT(*) AS total_orders,
        SUM(quantity) AS total_ordered_quantity
        FROM orders
        GROUP BY customer_id;""")
```

```python
# Bonus Challege
run_sql("""SELECT category,
     MIN(price) AS minimum_price,
     MAX(price) AS maximum_price
     FROM products
     GROUP BY category;""")
```
