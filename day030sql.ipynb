```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: Final review of ORDER,LIKE,%, and_.
```

```python
#1 Display all customers whose customer_name starts with 'A', sorted apphabetically.
run_sql("""SELECT *
FROM customers
WHERE first_name LIKE 'A%'
ORDER BY first_name ASC;""")
```

```python
#2 Display employees whose first_name contains the letter 'a', sorted by first_name.
run_sql("""SELECT *
FROM employees
WHERE first_name LIKE '%a%'
ORDER BY first_name ASC;""")
```

```python
#3 Display products whose product_name ends with 'r', sorted by price indescending order.
run_sql("""SELECT *
FROM products
WHERE product_name LIKE '%r'
ORDER BY price DESC;""")
```

```python
# Display department whose department_name contains the letter 'e', sorted alphabetically.
run_sql("""SELECT *
FROM departments
WHERE department_name LIKE '%e%'
ORDER BY department_name ASC;""")
```

```python
#5 Display customers whose city starts with 'B' and ends with 'e', sorted by customer_name.
run_sql("""SELECT *
FROM customers
WHERE city LIKE 'B%e'
ORDER BY first_name ASC;""")
```
