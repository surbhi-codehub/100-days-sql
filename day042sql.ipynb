```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
conn = sqlite3.connect('/content/company.db')

# Helper function to run SQL queries
def run_sql(query):
    return pd.read_sql_query(query, conn)

print("✅ Connected to company.db successfully!")
```

```python
# Topic: GROUP BY Multiple Aggregate Functions.
```

```python
# 1. Display each dapartment_id along with the total number of employees.
run_sql("""SELECT department_id, COUNT(*) AS total_employees
          FROM employees
          GROUP BY department_id;""")
```

```python
# 2. Display each product category along with the number of products and the average product price.
run_sql("""SELECT category,
         count(*) AS total_products,
         AVG(price) AS average_price
         FROM products
         GROUP BY category;""")
```

```python
# 3. Display each customer along with the total number of orders and the total orders and the total ordered quantity.
run_sql("""SELECT c.customer_id,
            COUNT(o.order_id) AS total_orders,
            SUM(o.quantity) AS total_ordered_quantity
            FROM customers c
            LEFT JOIN orders o
            ON c.customer_id = o.customer_id
            GROUP BY c.customer_id;""")
```

```python
# 4. Display each  department_id with the minimum salary, maximum salary, and average salary.
run_sql("""SELECT department_id,
        MIN(salary) AS minimum_salary,
        MAX(salary) AS maximum_salary,
        AVG(salary) AS average_salary
        FROM employees
        GROUP BY department_id;""")
```

```python
# 5. Display each product category with the total stock quantity, average product price, lowest price, and highest price.
run_sql(("""SELECT category,
SUM(stock_quantity) AS total_stock_qunatity,
AVG(price) AS average_product_price,
MIN(price) AS lowest_price,
MAX(price) AS highest_price
FROM products
GROUP BY category;"""))
```

```python
# Bonus Challege
run_sql("""SELECT customer_id,
        MIN(quantity) AS minimum_order_quantity,
        MAX(quantity) AS maximum_order_quantity
        FROM orders
        GROUP BY customer_id;""")
```
